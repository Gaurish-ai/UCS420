```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print(arr + 2)
print(arr * 3)
print(arr / 2)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print(arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
x_mode = np.bincount(x).argmax()
x_indices = np.where(x == x_mode)[0]
print(x_mode)
print(x_indices)

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
y_mode = np.bincount(y).argmax()
y_indices = np.where(y == y_mode)[0]
print(y_mode)
print(y_indices)
```

```python
arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print(arr[0, 1])
print(arr[2, 0])
```

```python
Gaurish = np.linspace(10, 100, 25)

print(Gaurish)
print(Gaurish.ndim)
print(Gaurish.shape)
print(Gaurish.size)
print(Gaurish.dtype)
print(Gaurish.nbytes)

transpose_reshape = Gaurish.reshape(25, 1)
print(transpose_reshape)

print(Gaurish.T)
```

```python
ucs420_Gaurish = np.array([[10, 20, 30, 40],
                          [50, 60, 70, 80],
                          [90, 15, 20, 35]])

print(np.mean(ucs420_Gaurish))
print(np.median(ucs420_Gaurish))
print(np.max(ucs420_Gaurish))
print(np.min(ucs420_Gaurish))
print(np.unique(ucs420_Gaurish))

reshaped_ucs420_Gaurish = ucs420_Gaurish.reshape(4, 3)
print(reshaped_ucs420_Gaurish)

resized_ucs420_Gaurish = np.resize(ucs420_Gaurish, (2, 3))
print(resized_ucs420_Gaurish)
```
