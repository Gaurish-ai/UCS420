## Q1. Text preprocessing and word frequency

The paragraph below is about artificial intelligence and technology.

```python
import re
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('stopwords', quiet=True)

paragraph = "Artificial intelligence is changing the way people learn, work, and solve problems. I enjoy exploring new technologies because they turn creative ideas into useful tools. Machine learning helps computers discover patterns from data and make better predictions. Natural language processing allows computers to understand and generate human language. I find this field exciting because it combines programming, mathematics, and creativity. In the future, responsible AI can make education, healthcare, and everyday life more efficient."

clean_text = re.sub(r'[^a-zA-Z0-9\\s]', '', paragraph.lower())
print("Original paragraph:\\n", paragraph)
print("\\nLowercase text without punctuation:\\n", clean_text)
```

```python
sentences = sent_tokenize(paragraph)
words = word_tokenize(clean_text)

print("Sentences:")
for sentence in sentences:
    print(sentence)

print("\\nWord tokens:")
print(words)
```

```python
python_split = clean_text.split()
nltk_tokens = word_tokenize(clean_text)

print("Python split() tokens:")
print(python_split)
print("\\nNLTK word_tokenize() tokens:")
print(nltk_tokens)
print("\\nNumber of tokens from split():", len(python_split))
print("Number of tokens from word_tokenize():", len(nltk_tokens))
print("\\nComparison: split() separates text wherever whitespace occurs. word_tokenize() uses language-aware tokenization rules and can separate punctuation and contractions into individual tokens. Since punctuation was removed before tokenization here, the outputs are very similar.")
```

```python
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

print("Words after stopword removal:")
print(filtered_words)
```

```python
frequency_distribution = FreqDist(filtered_words)
print("Word frequency distribution (excluding stopwords):")
for word, frequency in frequency_distribution.most_common():
    print(f"{word}: {frequency}")
```

## Q2. Alphabetic words, stemming and lemmatization

```python
alphabetic_words = re.findall(r'\\b[a-zA-Z]+\\b', paragraph.lower())
alphabetic_without_stopwords = [word for word in alphabetic_words if word not in stop_words]

print("Alphabetic words:")
print(alphabetic_words)
print("\\nAlphabetic words after stopword removal:")
print(alphabetic_without_stopwords)
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

porter = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stemmed_words = [porter.stem(word) for word in alphabetic_without_stopwords]
lemmatized_words = [lemmatizer.lemmatize(word) for word in alphabetic_without_stopwords]

print("Stemmed words:")
print(stemmed_words)
print("\\nLemmatized words:")
print(lemmatized_words)
```

```python
comparison = list(zip(alphabetic_without_stopwords, stemmed_words, lemmatized_words))
print(f"{'Original':<18}{'Stemmed':<18}{'Lemmatized':<18}")
print("-" * 54)
for original, stemmed, lemmatized in comparison:
    print(f"{original:<18}{stemmed:<18}{lemmatized:<18}")

print("\\nComparison and preference:")
print("Porter stemming removes or modifies word endings using rules, so its results may not be complete dictionary words. Lemmatization converts words to their meaningful base forms using vocabulary and linguistic rules. Stemming is useful when speed is important and exact word meaning is less important, such as basic search indexing. Lemmatization is preferable when readable and linguistically meaningful words are needed, such as in text analysis and language applications.")
```

## Q3. Bag of Words and TF-IDF

Three short texts are used: a technology news headline, a product review, and a second technology headline.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

texts = [
    "Artificial intelligence helps hospitals detect diseases earlier",
    "This smartphone has an excellent camera and long battery life",
    "New machine learning tools make data analysis faster"
]

for index, text in enumerate(texts, start=1):
    print(f"Text {index}: {text}")
```

```python
count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)
bow_features = count_vectorizer.get_feature_names_out()

print("Bag of Words vocabulary:")
print(bow_features)
print("\\nBag of Words matrix:")
print(bow_matrix.toarray())
```

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)
tfidf_features = tfidf_vectorizer.get_feature_names_out()

print("TF-IDF vocabulary:")
print(tfidf_features)
print("\\nTF-IDF matrix:")
print(tfidf_matrix.toarray().round(3))
```

```python
print("Top 3 TF-IDF keywords from each text:")
for index, row in enumerate(tfidf_matrix.toarray(), start=1):
    top_indices = row.argsort()[::-1][:3]
    top_keywords = [(tfidf_features[i], round(row[i], 3)) for i in top_indices]
    print(f"Text {index}: {top_keywords}")
```

## Interpretation

- **Bag of Words:** Represents each text using word counts. It records how often each vocabulary word appears, but does not consider word order or the importance of a word across documents.
- **TF-IDF:** Gives higher weight to words that are frequent in one text but less common across the collection. Common words across multiple texts receive lower importance.
- **Top keywords:** The three highest TF-IDF terms summarize the most distinctive content of each text. The keywords can help identify the main subject of a document.
