```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("a.", corrected_temperature)

fahrenheit = 9/5 * corrected_temperature + 32
print("b.", fahrenheit)

readings_greater_32 = corrected_temperature[corrected_temperature > 32]
print("c.", readings_greater_32)

count_greater_32 = np.sum(corrected_temperature > 32)
print("d.", count_greater_32)

print("e. Vectorization applies operations to the complete array at once, while Boolean indexing selects values based on a condition without explicitly iterating through each element.")
```

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("a.", np.sum(steps))
print("b.", np.mean(steps))
print("c. Maximum:", np.max(steps))
print("   Minimum:", np.min(steps))
print("d.", np.sum(steps, axis=0))
print("e.", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("f.", max_position)
```

```python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999
print("c. Original:", original)
print("   Subset:", subset)

copied = original[1:5].copy()
copied[0] = 500
print("d. Original:", original)
print("   Copied:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("e.", matrix)

print("f. First row:", matrix[0])
print("   Last row:", matrix[-1])
print("   Second column:", matrix[:, 1])
print("   Rows 1–2 and columns 2–3:", matrix[1:3, 2:4])

flattened = matrix.flatten()
raveled = matrix.ravel()
print("g. Flatten:", flattened)
print("   Ravel:", raveled)

raveled[0] = 100
print("h. Matrix after modifying ravel:", matrix)
print("   Ravel:", raveled)

flattened[1] = 200
print("i. Matrix after modifying flatten:", matrix)
print("   Flatten:", flattened)

print("j. Shape:", matrix.shape)
print("   Dimensions:", matrix.ndim)
print("   Size:", matrix.size)
print("   Dtype:", matrix.dtype)
```

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("a. Shape:", X.shape)
print("   Dimensions:", X.ndim)

XT = X.T
print("b. X.T:")
print(XT)
print("The transpose changes rows into columns, so each feature becomes a row.")

XTX = X.T @ X
print("c. X.T @ X:")
print(XTX)

XTX_inv = np.linalg.inv(XTX)
print("d. Inverse:")
print(XTX_inv)

beta = XTX_inv @ X.T @ y
print("e. Coefficients:", beta)

print("f. The coefficients represent the estimated change in assistance score associated with one-unit change in sleep hours, activity level, and stress level, while the other features are held constant.")

new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("g. Predicted assistance score:", prediction)
```
